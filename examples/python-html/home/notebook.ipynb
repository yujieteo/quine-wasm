```python
import numpy as np
x = np.linspace(0, 2 * np.pi, 9)
np.round(np.sin(x), 3)
```

```python
try:
    import matplotlib.pyplot as plt
except ImportError:
    print("matplotlib is in python-sci.html")
else:
    plt.plot(x, np.sin(x), marker="o")
    plt.savefig("sine.svg")  # a file in the user folder: Save keeps it
```
